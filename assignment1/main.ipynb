## 2. Знакомство с данными

### Общая структура данных

Набор данных содержит 193 наблюдения и 10 признаков. Каждая строка соответствует отдельной стране или территории, а признаки описывают показатель среднего IQ, континент, уровень грамотности, количество Нобелевских премий, индекс человеческого развития (HDI), среднюю продолжительность обучения, GNI и численность населения.

Признаки имеют разные типы. `Country` и `Continent` являются категориальными переменными, `Rank` представляет собой порядковый ранг, а остальные признаки в основном являются количественными.

Большинство признаков заполнены полностью. Однако для трёх показателей, связанных с HDI, наблюдается по 179 значений из 193, то есть по 14 пропусков в каждом признаке. Это составляет приблизительно 7.3% наблюдений и потребует отдельного рассмотрения на этапе предобработки.

Также `Population - 2023` представлена в формате `str`, хотя по смыслу является числовым показателем. Перед использованием этого признака в математических операциях его потребуется преобразовать в числовой тип.

### Первичные статистические характеристики

Показатель `Average IQ` имеет среднее значение 82.05 и медианное значение 82.24. Они находятся близко друг к другу, однако этого недостаточно для вывода о симметричности распределения; его форма будет дополнительно исследована с помощью визуализаций.

Диапазон значений `Average IQ` составляет от 42.99 до 106.48, а стандартное отклонение равно 13.34, что указывает на заметный разброс значений между наблюдениями.

`Literacy Rate` принимает значения от 0.19 до 1.00. При этом медиана равна 0.95, а 75-й процентиль — 0.99, поэтому распределение этого показателя также представляет интерес для дальнейшего анализа.

Для `GNI - 2021` наблюдается значительный разброс: значения находятся в диапазоне от 732 до 146830. Стандартное отклонение составляет 22160, что сравнимо и даже несколько превышает среднее значение 20812. Это может указывать на существенную неоднородность показателя.

Особенно заметным является распределение `Nobel Prices`: медианное значение равно 0, 75-й процентиль — 1, тогда как максимальное значение достигает 400. Таким образом, этот признак имеет крайне неравномерное распределение и требует дополнительного изучения перед использованием в сравнительном анализе.

### Форматирование численности населения

При первичном анализе было обнаружено, что признак `Population - 2023`
сохранён как строковый тип (`str`). Дополнительная проверка показала наличие
лишних пробельных символов в некоторых значениях, например табуляции перед
числом.

Для корректного использования признака в численных расчётах были удалены
лишние пробельные символы, после чего значения преобразованы в целочисленный
тип.

```python
for i, value in df["Population - 2023"].items():
    try:
        int(value.strip())
    except ValueError:
        print(i, repr(value))
```

```python
df["Population - 2023"] = (
    df["Population - 2023"]
    .str.strip()
    .str.replace(".", "", regex=False)
    .astype(int)
)

df["Population - 2023"].dtype
```

## 4. Основной анализ и визуализация

В данном разделе проводится основной разведывательный анализ данных. Для каждой визуализации формулируется отдельный исследовательский вопрос, после чего полученный результат интерпретируется с учётом особенностей набора данных.

Первым этапом рассмотрим распределение показателя `Average IQ` среди представленных наблюдений.

### 4.1. Распределение среднего IQ

**Исследовательский вопрос:** как распределены значения `Average IQ` среди стран, представленных в наборе данных?

**Предположение:** значения могут быть распределены неравномерно, поэтому интересно определить центральную область распределения, его разброс и наличие потенциальных выбросов.

```python
fig = px.histogram(
    df,
    x="Average IQ",
    nbins=20,
    title="average iq distribution",
    labels={
        "Average IQ": "Average IQ score",
        "count": "The number of countries",
    }
)

fig.update_layout(
    xaxis_title="Average IQ score",
    yaxis_title="The number of countries",
)

fig.show()
```

### 4.2. Распределение IQ по континентам

**Исследовательский вопрос:** различается ли распределение показателя `Average IQ` между континентами?

**Предположение:** значения `Average IQ` могут различаться между континентами, поэтому сравним их медианы, разброс и потенциальные выбросы.

```python
fig = px.box(
    df,
    x="Continent",
    y="Average IQ",
    color="Continent",
    points="all",
    title="average iq distribution by continent",
    labels={
        "Continent": "Continent",
        "Average IQ": "Average IQ score"
    }
)

fig.update_layout(
    xaxis_title="Continent",
    yaxis_title="Average IQ",
    showlegend=False
)

fig.show()
```

### 4.3. Связь между IQ и индексом человеческого развития

**Исследовательский вопрос:** существует ли визуально заметная связь между `Average IQ` и `HDI (2021)`?

**Предположение:** страны с более высоким индексом человеческого развития могут иметь более высокие значения `Average IQ`. Проверим, наблюдается ли такая тенденция в представленных данных.

```python
fig = px.scatter(
    df,
    x="HDI (2021)",
    y="Average IQ",
    color="Continent",
    hover_name="Country",
    hover_data={
        "HDI (2021)": True,
        "Average IQ": True,
        "Continent": True,
    },
    title="relationship between IQ and HDI",
    labels={
        "HDI (2021)": "Human Development Index",
        "Average IQ": "Average IQ",
        "Continent": "Continent"
    }
)

fig.update_layout(
    xaxis_title="Human Development Index",
    yaxis_title="Average IQ"
)

fig.show()
```

### 4.4. Связь между IQ и продолжительностью обучения

**Исследовательский вопрос:** существует ли связь между `Average IQ` и средней продолжительностью обучения населения?

**Предположение:** страны, в которых средняя продолжительность обучения выше, могут иметь более высокие значения `Average IQ`. Проверим, наблюдается ли подобная тенденция в данных.

```python
indicators = {
    "HDI (2021)": "HDI",
    "Mean years of schooling - 2021": "Mean years of schooling",
    "Literacy Rate": "Literacy Rate",
    "GNI - 2021": "GNI"
}

dropdown = widgets.Dropdown(
    options=indicators.keys(),
    value="Mean years of schooling - 2021",
    description="Metric:",
    style={"description_width": "initial"}
)

def update_plot(indicator):
    plot_df = df.dropna(subset=[indicator, "Average IQ"])

    fig = px.scatter(
        plot_df,
        x=indicator,
        y="Average IQ",
        color="Continent",
        hover_name="Country",
        hover_data={
            indicator: True,
            "Average IQ": True,
            "Continent": True
        },
        title=f"relation between IQ and {indicators[indicator]}",
        labels={
            indicator: indicators[indicator],
            "Average IQ":" Average IQ",
            "Continent": "Continent"
        }
    )

    fig.update_layout(
        xaxis_title=indicators[indicator],
        yaxis_title="Average IQ",
        showlegend=True,
    )

    fig.show()

widgets.interact(
    update_plot,
    indicator=dropdown
);
```
